```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Generate 2 years of daily dates
dates = pd.date_range(start="2024-01-01", end="2025-12-31", freq="D")

# (average daily patients, supply cost per patient in INR, patients per nurse)
departments = {
    "Emergency": (45, 4200, 4),
    "ICU": (18, 14500, 2),
    "General Ward": (70, 2500, 8),
    "Pediatrics": (30, 3100, 5),
}

rows = []

for date in dates:
    day_of_year = date.timetuple().tm_yday
    # Simulate seasonal peaks (higher admissions during monsoon/winter)
    seasonal_wave = 1 + 0.20 * np.sin(2 * np.pi * (day_of_year - 180) / 365)
    is_weekend = 1 if date.weekday() >= 5 else 0

    for dept, (base_patients, unit_cost, ratio) in departments.items():
        weekend_factor = 1.15 if (dept == "Emergency" and is_weekend) else (0.90 if is_weekend else 1.0)
        random_noise = np.random.normal(1.0, 0.08)

        admissions = max(1, int(round(base_patients * seasonal_wave * weekend_factor * random_noise)))
        nurses_needed = int(np.ceil(admissions / ratio))
        doctors_needed = max(1, int(np.ceil(nurses_needed / 3)))

        staff_cost = (nurses_needed * 1800) + (doctors_needed * 5500)
        supply_cost = int(admissions * unit_cost * np.random.uniform(0.92, 1.08))
        overhead_cost = int(base_patients * 600)
        total_budget = staff_cost + supply_cost + overhead_cost

        rows.append({
            "date": date.strftime("%Y-%m-%d"),
            "department": dept,
            "day_of_week": date.dayofweek,
            "month": date.month,
            "is_weekend": is_weekend,
            "admissions": admissions,
            "nurses_needed": nurses_needed,
            "doctors_needed": doctors_needed,
            "staff_cost": staff_cost,
            "supply_cost": supply_cost,
            "total_budget": total_budget,
        })

df = pd.DataFrame(rows)
df.to_csv("hospital_operations.csv", index=False)

print("Success! Created hospital_operations.csv")
print("Total rows:", len(df))
print(df.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("hospital_operations.csv", parse_dates=["date"])

print("Shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
display(df.head())
display(df.describe().round(1))
```

```python
monthly_adm = df.groupby(["month", "department"])["admissions"].mean().unstack()

monthly_adm.plot(figsize=(10, 5), marker="o")
plt.title("Average Daily Admissions by Month and Department")
plt.xlabel("Month (1 = Jan, 12 = Dec)")
plt.ylabel("Average Daily Admissions")
plt.xticks(range(1, 13))
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend(title="Department")
plt.tight_layout()
plt.savefig("monthly_admissions.png", dpi=150)
plt.show()
```

```python
dept_costs = df.groupby("department")[["staff_cost", "supply_cost"]].mean()
dept_costs["overhead_cost"] = (
    df.groupby("department")["total_budget"].mean()
    - dept_costs["staff_cost"]
    - dept_costs["supply_cost"]
)

# Convert INR to Lakhs for cleaner axis labels
dept_costs_lakhs = dept_costs / 100000

dept_costs_lakhs.plot(kind="bar", stacked=True, figsize=(9, 5))
plt.title("Average Daily Budget Breakdown by Department")
plt.xlabel("Department")
plt.ylabel("Average Daily Cost (INR Lakhs)")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend(["Staff Cost", "Supply Cost", "Overhead Cost"])
plt.tight_layout()
plt.savefig("department_budget_breakdown.png", dpi=150)
plt.show()
```

```python
weekend_cmp = df.groupby(["department", "is_weekend"])["admissions"].mean().unstack()
weekend_cmp.columns = ["Weekday", "Weekend"]

weekend_cmp.plot(kind="bar", figsize=(9, 5))
plt.title("Weekday vs. Weekend Average Daily Admissions")
plt.xlabel("Department")
plt.ylabel("Average Daily Admissions")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("weekend_vs_weekday.png", dpi=150)
plt.show()
```

```python
df["year_month"] = df["date"].dt.to_period("M")
monthly_budget = df.groupby("year_month")["total_budget"].sum() / 1e7  # INR Crores

monthly_budget.plot(figsize=(11, 4), marker="o", color="teal")
plt.title("Total Monthly Hospital Budget (2024–2025)")
plt.xlabel("Month")
plt.ylabel("Total Budget (INR Crores)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.savefig("monthly_total_budget.png", dpi=150)
plt.show()
```

```python
%%writefile app.py
from datetime import date
import os
import joblib
import numpy as np
import pandas as pd
import streamlit as st

st.set_page_config(
    page_title="Hospital Resource & Budget Forecaster",
    layout="wide",
)

DEPARTMENTS = ["Emergency", "ICU", "General Ward", "Pediatrics"]


@st.cache_resource
def load_model():
    if not os.path.exists("model.pkl"):
        return None
    return joblib.load("model.pkl")


def format_inr(amount: float) -> str:
    if amount >= 1e7:
        return f"₹{amount / 1e7:.2f} Cr"
    if amount >= 1e5:
        return f"₹{amount / 1e5:.2f} Lakhs"
    return f"₹{amount:,.0f}"


def generate_forecast(model, start_date, days: int, selected_dept: str, surge_pct: int) -> pd.DataFrame:
    dates = pd.date_range(start=start_date, periods=days, freq="D")
    depts = DEPARTMENTS if selected_dept == "All Departments" else [selected_dept]

    rows = []
    for d in dates:
        for dept in depts:
            rows.append({
                "date": d,
                "department": dept,
                "day_of_year": d.dayofyear,
                "month": d.month,
                "day_of_week": d.dayofweek,
                "is_weekend": 1 if d.dayofweek >= 5 else 0,
            })

    input_df = pd.DataFrame(rows)
    feature_cols = ["department", "day_of_year", "month", "day_of_week", "is_weekend"]
    preds = model.predict(input_df[feature_cols])

    multiplier = 1 + (surge_pct / 100.0)

    result = pd.DataFrame({
        "Date": input_df["date"].dt.strftime("%Y-%m-%d"),
        "Department": input_df["department"],
        "Predicted Admissions": np.round(preds[:, 0] * multiplier).astype(int),
        "Nurses Needed": np.ceil(preds[:, 1] * multiplier).astype(int),
        "Doctors Needed": np.ceil(preds[:, 2] * multiplier).astype(int),
        "Estimated Budget (INR)": np.round(preds[:, 3] * multiplier).astype(int),
    })
    return result


# --- Main UI ---
st.title("Healthcare Resource Allocation & Budget Forecasting")
st.caption("AI-driven patient volume, staffing, and operational budget planner (SDG 3)")

model = load_model()
if model is None:
    st.error("`model.pkl` not found in this folder. Run `train_model.py` first.")
    st.stop()

# Sidebar controls
st.sidebar.header("Forecast Controls")
selected_dept = st.sidebar.selectbox(
    "Department",
    ["All Departments"] + DEPARTMENTS,
)
start_date = st.sidebar.date_input("Start Date", value=date.today())
forecast_days = st.sidebar.slider("Forecast Horizon (Days)", min_value=1, max_value=90, value=30)
surge_pct = st.sidebar.slider(
    "Surge / Contingency Buffer (%)",
    min_value=0,
    max_value=30,
    value=0,
    help="Simulate seasonal outbreaks or supply cost inflation.",
)

forecast_df = generate_forecast(model, start_date, forecast_days, selected_dept, surge_pct)

# Aggregate daily totals across selected departments
daily_totals = (
    forecast_df.groupby("Date", as_index=False)[
        ["Predicted Admissions", "Nurses Needed", "Doctors Needed", "Estimated Budget (INR)"]
    ]
    .sum()
)

# KPI Metric Cards
col1, col2, col3, col4 = st.columns(4)
col1.metric("Total Forecasted Admissions", f"{int(daily_totals['Predicted Admissions'].sum()):,}")
col2.metric("Peak Daily Nurses Needed", f"{int(daily_totals['Nurses Needed'].max())}")
col3.metric("Peak Daily Doctors Needed", f"{int(daily_totals['Doctors Needed'].max())}")
col4.metric("Total Estimated Budget", format_inr(daily_totals["Estimated Budget (INR)"].sum()))

st.divider()

# Trend Charts
chart_col1, chart_col2 = st.columns(2)
with chart_col1:
    st.subheader("Daily Predicted Admissions")
    st.line_chart(daily_totals.set_index("Date")["Predicted Admissions"])

with chart_col2:
    st.subheader("Daily Estimated Budget (INR)")
    st.line_chart(daily_totals.set_index("Date")["Estimated Budget (INR)"])

# Department Breakdown (only shown when viewing all departments)
if selected_dept == "All Departments":
    st.subheader("Department Budget & Resource Breakdown")
    dept_summary = (
        forecast_df.groupby("Department", as_index=False)
        .agg(
            Total_Admissions=("Predicted Admissions", "sum"),
            Peak_Daily_Nurses=("Nurses Needed", "max"),
            Peak_Daily_Doctors=("Doctors Needed", "max"),
            Total_Budget_INR=("Estimated Budget (INR)", "sum"),
        )
    )
    b_col1, b_col2 = st.columns([1.2, 1])
    with b_col1:
        st.bar_chart(dept_summary.set_index("Department")["Total_Budget_INR"])
    with b_col2:
        st.dataframe(dept_summary, hide_index=True)

# Detailed Schedule Table + CSV Export
st.subheader("Detailed Daily Forecast")
st.dataframe(forecast_df, hide_index=True)

csv_data = forecast_df.to_csv(index=False).encode("utf-8")
st.download_button(
    label="Download Forecast as CSV",
    data=csv_data,
    file_name=f"hospital_forecast_{start_date}_{forecast_days}d.csv",
    mime="text/csv",
)
```
